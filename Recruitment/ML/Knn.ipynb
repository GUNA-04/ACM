```python

import matplotlib.pyplot as plt
import numpy as np
```

```python
k =3
orange = np.array([(1,3), (2,4)])
yellow = np.array([(2,3),(2,2)])
```

```python
test = np.array([1,6])
x_train = np.concatenate((orange,yellow), axis=0)
y_train = np.array([0,0,1,1])

distance = np.linalg.norm(x_train-test, axis=1)
```

```python
nearest = distance.argsort()[:k]
nearest_labels = y_train[nearest]
count = np.bincount(nearest_labels)
ans= count.argmax()
```

```python
print("Predicted Class:", "Orange" if ans== 0 else "Yellow")
```
