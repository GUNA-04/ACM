```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as  plt
from sklearn.model_selection import train_test_split    

from sklearn.cluster import KMeans

```

```python
data = pd.read_csv("C:\\Users\\techn\\Documents\\GitHub\\ACM\\Recruitment\\ML\\Unsupervised\\sales.csv")
print(data.head())
print(data.info())
print(data.isnull().sum() )
```

```python
for colum in data.columns:
    if data[colum].dtype == 'object':
        data[colum] = pd.Categorical(data[colum]).codes 
```

```python
train_test_split(data, test_size=0.8, random_state=42)
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(data)
data['cluster'] = kmeans.labels_
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(data['Price'], data['Quantity Sold'], c=data['cluster'], cmap='viridis', s=100, edgecolor='k')
plt.xlabel('Price')
plt.ylabel('Quantity Sold')
plt.title('K-Means Clustering')
plt.grid(True)
plt.show()
```
