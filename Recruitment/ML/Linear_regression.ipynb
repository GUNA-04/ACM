```python

import matplotlib.pyplot as plt 
import numpy as np
```

```python
x = np.array([1,2,3,4,5])
y = np.array([4,5,6,7,8])
```

```python
m= 0;
c =0;
alpha = 0.01
repeat = 5000
n = len(x)
```

```python
for i in range(repeat):
    y_predicted = m*x+c
    error = y_predicted - y
    cost = (1/n) * np.sum(error * error)
```

```python
dm = (2/n)*np.sum(error*x)
dc = (2/n)*np.sum(error)
m= m-alpha*dm
c= c- alpha*dc
```
